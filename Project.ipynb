# Descriptive Analysis
## For quantitative data :
- mean (expected value)
- standard deviation (how much data is far from the mean)
- skewness : 0 means totally centered (on the mean), <0 means more small values,...
- Kurtosis : measures the height of the peak (0 is like a normal, python function is normalized)

```python
stats_dict = {}
for variable in df.columns :
    if (df.dtypes[variable] == "int64" or df.dtypes[variable] == "float64"):
        stats_dict[variable] = {
            "mean":df[variable].mean(),
            "std":df[variable].std(),
            "skew":df[variable].skew(),
            "kurt":df[variable].kurt()
        } 
        print(f"{variable} {stats_dict[variable]}")
    else :
        # Est-ce qu'on discrétise les variables ou on les laisse comme ca et fait d'autres valeurs ?
        pass
```
